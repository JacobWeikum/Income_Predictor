```python
# Importing key tools
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Importing data
depression_data = pd.read_csv("data/depression_data.csv")
depression_data.head()
```

```python
# Checking for missing data
depression_data.isna().sum()
```

```python
# Finding data types
depression_data.dtypes
```

```python
# Seperating x and y variables
x = depression_data.drop("Income", axis=1)
y = depression_data["Income"]

# Dropping Name for ease of training
x = x.drop("Name", axis=1)

x.head()
```

```python
# Converting objects into numeric data with and then categorizing it
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ["Marital Status", "Education Level", "Smoking Status", "Physical Activity Level", 
                        "Employment Status", "Alcohol Consumption", "Dietary Habits", "Sleep Patterns",
                        "History of Mental Illness", "History of Substance Abuse", "Family History of Depression",
                        "Chronic Medical Conditions"]
one_hot = OneHotEncoder()
transformed_x = ColumnTransformer([("one_hot",
                                 one_hot,
                                 categorical_features)],
                                remainder="passthrough").fit_transform(x)
transformed_x
```

```python
# Trainging transformed_x with RandomForestRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, Y_test = train_test_split(transformed_x, y, test_size=0.2)
model = RandomForestRegressor()
model.fit(x_train, y_train)
```

```python
model.score(x_test, Y_test)
```
